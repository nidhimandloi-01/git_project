```python
import pandas as pd

df = pd.read_csv("zomato.csv", encoding="utf-8", engine="python", on_bad_lines="warn")
print(df.shape)
print(df.columns.tolist())
```

```python
df = df.rename(columns={
    "name": "restaurant_name",
    "book_table": "table_booking",
    "rate": "rating",
    "approx_cost(for two people)": "cost_for_two",
    "listed_in(type)": "listing_type",
    "listed_in(city)": "listing_city",
})
```

```python
def fix_text(s):
    if not isinstance(s, str):
        return s
    s = s.strip()
    try:  # fixes mojibake like "CafÃ©" -> "Café"
        return s.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return s

text_cols = ["restaurant_name", "address", "location", "rest_type", "cuisines",
             "listing_type", "listing_city", "online_order", "table_booking"]
for c in text_cols:
    df[c] = df[c].map(fix_text)
```

```python
raw_rate = df["rating"].astype(str).str.strip()

def clean_rate(x):
    x = str(x).strip()
    if x in ("NEW", "-", "", "nan"):
        return np.nan
    try:
        return float(x.split("/")[0])
    except ValueError:
        return np.nan

df["rating"] = raw_rate.map(clean_rate)
df["rating_status"] = np.where(
    raw_rate.eq("NEW"), "New",
    np.where(df["rating"].notna(), "Rated", "Not rated"),
)
```

```python
df["cost_for_two"] = pd.to_numeric(
    df["cost_for_two"].astype(str).str.replace(",", "", regex=False),
    errors="coerce",
)
```

```python
df["votes"] = pd.to_numeric(df["votes"], errors="coerce").fillna(0).astype(int)
```

```python
for c in ["location", "rest_type", "cuisines"]:
    df[c] = df[c].fillna("Unknown")

# Restaurant name ke bina row useless hai
df = df.dropna(subset=["restaurant_name"])
```

```python
before = len(df)
df = df.drop_duplicates()
print("Duplicates removed:", before - len(df))
```

```python
df["restaurant_id"] = pd.factorize(
    df["restaurant_name"] + "|" + df["address"].fillna("")
)[0] + 1
```

```python
df["rating_band"] = (
    pd.cut(df["rating"], bins=[0, 2.5, 3.0, 3.5, 4.0, 5.0],
           labels=["2.5 or below", "2.6-3.0", "3.1-3.5", "3.6-4.0", "4.1-5.0"])
    .astype(str).replace("nan", "Unrated")
)
df["cost_band"] = (
    pd.cut(df["cost_for_two"], bins=[0, 300, 600, 1000, 2000, np.inf],
           labels=["Under 300", "300-600", "600-1000", "1000-2000", "2000+"])
    .astype(str).replace("nan", "Unknown")
)
df["primary_rest_type"] = df["rest_type"].str.split(",").str[0].str.strip()
df["primary_cuisine"] = df["cuisines"].str.split(",").str[0].str.strip()
df["cuisine_count"] = df["cuisines"].str.split(",").str.len()
df.loc[df["cuisines"].eq("Unknown"), "cuisine_count"] = 0
```

```python
final_cols = [
    "restaurant_id", "restaurant_name", "online_order", "table_booking",
    "rating", "rating_status", "rating_band", "votes",
    "cost_for_two", "cost_band",
    "location", "listing_city", "listing_type",
    "rest_type", "primary_rest_type", "cuisines", "primary_cuisine", "cuisine_count",
]
clean = df[final_cols].sort_values(["restaurant_id", "listing_type"]).reset_index(drop=True)
```

```python
bridge = (
    clean[["restaurant_id", "cuisines"]]
    .drop_duplicates("restaurant_id")
    .assign(cuisine=lambda d: d["cuisines"].str.split(","))
    .explode("cuisine")
)
bridge["cuisine"] = bridge["cuisine"].str.strip()
bridge = bridge[["restaurant_id", "cuisine"]].reset_index(drop=True)
```

```python
clean.to_csv("zomato_clean.csv", index=False, encoding="utf-8-sig")
bridge.to_csv("cuisines_bridge.csv", index=False, encoding="utf-8-sig")

# ---------- Quick sanity report ----------
print("\nClean shape:", clean.shape)
print("Unique restaurants:", clean["restaurant_id"].nunique())
print("\nNulls per column:\n", clean.isna().sum())
print("\nRating range:", clean["rating"].min(), "-", clean["rating"].max())
print("Cost range:", clean["cost_for_two"].min(), "-", clean["cost_for_two"].max())
```
